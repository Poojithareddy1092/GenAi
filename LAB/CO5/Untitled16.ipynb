```python
# Experiment 7: Hallucination Detection

reference = """
Python was created by Guido van Rossum.
Python was first released in 1991.
"""

questions = [
    "Who created Python?",
    "When was Python released?",
    "Was Python created in 1850?"
]

answers = [
    "Python was created by Guido van Rossum.",
    "Python was released in 1991.",
    "Python was created in 1850."
]

print("Reference Information:")
print(reference)

for q, a in zip(questions, answers):
    print("\nQuestion:", q)
    print("Answer:", a)

    if "1850" in a:
        print("Result: Hallucination detected")
    else:
        print("Result: Supported by reference")
```

```python
# Experiment 9: Simple RAG System

documents = [
    "Electric motors convert electrical energy into mechanical energy.",
    "Motor overheating occurs due to excessive load and poor ventilation.",
    "Batteries store electrical energy and require proper charging.",
    "Transformers transfer electrical energy between circuits."
]

question = input("Ask an engineering question: ")

words = question.lower().split()

scores = []

for doc in documents:
    score = sum(word in doc.lower() for word in words)
    scores.append(score)

best_index = scores.index(max(scores))

print("\nRetrieved Document:")

if max(scores) > 0:
    print(documents[best_index])
    print("\nGenerated Answer:")
    print(documents[best_index])
else:
    print("No relevant document found.")
```

```python
# Experiment 10: Engineering Troubleshooting System

documents = {
    "motor": "Motor overheating: Check the load, cooling fan and ventilation. Perform regular maintenance.",
    "battery": "Battery not charging: Check the power supply, charging cable and battery connections.",
    "transformer": "Transformer overheating: Check the electrical load and cooling system. Contact a qualified technician."
}

problem = input("Enter engineering problem: ").lower()

found = False

for key, solution in documents.items():

    if key in problem:
        print("\nRetrieved Technical Information:")
        print(solution)

        print("\nRecommended Steps:")
        for i, step in enumerate(solution.split(".")[:3], 1):
            if step.strip():
                print(i, ".", step.strip())

        found = True
        break

if not found:
    print("No matching technical document found.")
```
