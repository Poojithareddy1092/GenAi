```python
# Experiment 6: Text Generation Using OpenAI API

from openai import OpenAI
from getpass import getpass

api_key = getpass("Enter OpenAI API Key: ")
client = OpenAI(api_key=api_key)

prompt = input("Enter your prompt: ")

response = client.responses.create(
    model="gpt-5.5",
    input=prompt
)

print("\nGenerated Output:")
print(response.output_text)
```

```python
# Experiment 7: Text Generation Using Gemini API

from google import genai
from getpass import getpass

api_key = getpass("Enter Gemini API Key: ")
client = genai.Client(api_key=api_key)

while True:
    prompt = input("\nEnter your prompt (type exit to stop): ")

    if prompt.lower() == "exit":
        break

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )

    print("\nGenerated Output:")
    print(response.text)
```

```python
# Experiment 8: Text Generation Using Hugging Face API

from huggingface_hub import InferenceClient
from getpass import getpass

token = getpass("Enter Hugging Face Token: ")

client = InferenceClient(
    model="HuggingFaceH4/zephyr-7b-beta",
    token=token
)

prompt = input("Enter your prompt: ")

response = client.text_generation(
    prompt,
    max_new_tokens=150
)

print("\nGenerated Output:")
print(response)
```

```python
# Experiment 9: Structured Prompting for Python Code Generation

prompt = """
Generate a Python program using the following instructions:

Problem: Find the factorial of a number.

Input: An integer entered by the user.
Process: Calculate factorial using a loop.
Output: Display the factorial.

Requirements:
1. Use Python.
2. Write simple code.
3. Handle negative numbers.
4. Include proper output.
"""

print("STRUCTURED PROMPT:")
print(prompt)

# Sample generated Python program

code = """
n = int(input("Enter a number: "))

if n < 0:
    print("Factorial is not defined")
else:
    fact = 1
    for i in range(1, n + 1):
        fact = fact * i

    print("Factorial:", fact)
"""

print("\nGenerated Code:")
print(code)

# Validate the generated code

import ast

try:
    ast.parse(code)
    print("Code Validation: Successful")
except SyntaxError:
    print("Code Validation: Failed")
```

```python
# Experiment 10: SQL Query Generation Using Structured Prompting

prompt = """
Database: College Database

Table: Students

Columns:
id - Integer
name - Text
department - Text
marks - Integer

Requirement:
Generate an SQL query to display students
who scored more than 80 marks.

Output:
Write a valid SQL SELECT query.
"""

print("STRUCTURED PROMPT:")
print(prompt)

# Sample generated SQL query

query = """
SELECT * FROM Students
WHERE marks > 80;
"""

print("\nGenerated SQL Query:")
print(query)

# Validate SQL using SQLite

import sqlite3

conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE Students (
    id INTEGER,
    name TEXT,
    department TEXT,
    marks INTEGER
)
""")

cursor.execute("INSERT INTO Students VALUES (1, 'Arun', 'CSE', 90)")
cursor.execute("INSERT INTO Students VALUES (2, 'Ravi', 'ECE', 75)")
cursor.execute("INSERT INTO Students VALUES (3, 'Priya', 'CSE', 85)")

cursor.execute(query)

print("Query Result:")
for row in cursor.fetchall():
    print(row)

conn.close()
```

```python
# Experiment 11: Comparative Analysis of Prompting Techniques

task = "Explain the benefits of Artificial Intelligence in education."

zero_shot = task

one_shot = """
Example:
Topic: AI in Healthcare
Output: AI helps doctors diagnose diseases and improve patient care.

Task: Explain the benefits of Artificial Intelligence in education.
"""

few_shot = """
Example 1:
Topic: AI in Banking
Output: AI helps detect fraud and improve customer service.

Example 2:
Topic: AI in Agriculture
Output: AI helps farmers monitor crops and improve productivity.

Task: Explain the benefits of Artificial Intelligence in education.
"""

print("ZERO-SHOT PROMPT:")
print(zero_shot)

print("\nONE-SHOT PROMPT:")
print(one_shot)

print("\nFEW-SHOT PROMPT:")
print(few_shot)

print("\nCOMPARATIVE ANALYSIS")

comparison = {
    "Zero-shot": ["Good", "Moderate", "Simple", "Moderate"],
    "One-shot": ["Good", "Good", "Clear", "Good"],
    "Few-shot": ["High", "High", "Very Clear", "High"]
}

print("Technique | Quality | Relevance | Readability | Consistency")

for technique, values in comparison.items():
    print(technique, "|", " | ".join(values))
```

```python
# Experiment 12: Prompt Evaluation and Refinement

prompts = {
    "Prompt 1": "Explain Artificial Intelligence.",

    "Prompt 2": """
    Explain Artificial Intelligence in simple words.
    Include its definition and applications.
    """,

    "Prompt 3": """
    Explain Artificial Intelligence in simple words.
    Include its definition, applications, advantages,
    and examples. Write the answer in 150 words
    using clear headings and bullet points.
    """
}

# Evaluate prompts using predefined criteria

scores = {
    "Prompt 1": [6, 7, 5, 6, 7],
    "Prompt 2": [8, 8, 8, 8, 8],
    "Prompt 3": [9, 9, 9, 9, 10]
}

criteria = [
    "Relevance",
    "Accuracy",
    "Completeness",
    "Clarity",
    "Format Adherence"
]

for name, prompt in prompts.items():
    print("\n", name)
    print("Prompt:", prompt)

    for i in range(len(criteria)):
        print(criteria[i], ":", scores[name][i], "/10")

    print("Total Score:", sum(scores[name]), "/50")

# Select and refine the best prompt

best_prompt = """
Explain Artificial Intelligence in simple words.
Include its definition, applications, advantages,
and real-world examples. Write approximately 150 words
using clear headings and bullet points.
"""

print("\nREFINED PROMPT:")
print(best_prompt)

print("\nBest Prompt: Prompt 3")
print("Reason: It provides clear instructions and better output structure.")
```
