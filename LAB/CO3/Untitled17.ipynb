```python
# Experiment 1: Text Embeddings and Semantic Similarity Search

!pip install -q sentence-transformers

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

model = SentenceTransformer('all-MiniLM-L6-v2')

sentences = [
    "Artificial Intelligence enables machines to perform intelligent tasks.",
    "Machine Learning allows computers to learn from data.",
    "Deep Learning uses neural networks.",
    "Python is a popular programming language.",
    "Robotics combines software and mechanical systems."
]

embeddings = model.encode(sentences)

query = "How do computers learn from information?"
query_embedding = model.encode([query])

scores = cosine_similarity(query_embedding, embeddings)[0]

print("Query:", query)
print("\nSimilarity Results:")

for i in scores.argsort()[::-1]:
    print(sentences[i], "->", round(scores[i], 3))
```

```python
# Experiment 2: Semantic Search Using Cosine Similarity

!pip install -q sentence-transformers

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

model = SentenceTransformer('all-MiniLM-L6-v2')

documents = [
    "Solar energy is obtained from sunlight.",
    "Wind energy is generated using wind turbines.",
    "Electric vehicles use rechargeable batteries.",
    "Water pollution affects aquatic organisms.",
    "Artificial intelligence is used in healthcare."
]

query = "How is electricity produced from the sun?"

doc_embeddings = model.encode(documents)
query_embedding = model.encode([query])

similarity = cosine_similarity(query_embedding, doc_embeddings)[0]

best_index = similarity.argmax()

print("Search Query:", query)
print("\nMost Relevant Document:")
print(documents[best_index])
print("Similarity Score:", round(similarity[best_index], 3))
```

```python
# Experiment 3: Vector Database Using FAISS

!pip install -q sentence-transformers faiss-cpu

import faiss
import numpy as np
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')

documents = [
    "Python is used for programming.",
    "Machine learning is a branch of AI.",
    "Deep learning uses neural networks.",
    "Cloud computing provides online services.",
    "Computer vision helps machines understand images."
]

embeddings = model.encode(documents)
embeddings = np.array(embeddings).astype('float32')

index = faiss.IndexFlatL2(embeddings.shape[1])
index.add(embeddings)

query = "Explain artificial intelligence and learning."
query_embedding = model.encode([query]).astype('float32')

distances, results = index.search(query_embedding, 3)

print("Query:", query)
print("\nTop 3 Retrieved Documents:")

for i in results[0]:
    print("-", documents[i])
```

```python
# Experiment 4: Document Storage and Top-K Retrieval

!pip install -q sentence-transformers faiss-cpu

import faiss
import numpy as np
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')

documents = [
    "Artificial Intelligence is used in healthcare.",
    "Machine learning helps predict future outcomes.",
    "Python supports data science applications.",
    "Robotics is used in industrial automation.",
    "Cybersecurity protects computer networks.",
    "Cloud computing provides scalable storage."
]

embeddings = model.encode(documents).astype('float32')

database = faiss.IndexFlatL2(embeddings.shape[1])
database.add(embeddings)

query = "Applications of AI in medical fields"
query_vector = model.encode([query]).astype('float32')

k = 3
distances, indices = database.search(query_vector, k)

print("Query:", query)
print("\nTop", k, "Retrieved Documents:")

for rank, idx in enumerate(indices[0], 1):
    print(rank, documents[idx])
```

```python
# Experiment 5: Document Question Answering Using RAG

!pip install -q sentence-transformers

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

model = SentenceTransformer('all-MiniLM-L6-v2')

documents = [
    "Artificial Intelligence enables machines to perform human-like tasks.",
    "Machine Learning is a subset of AI that learns from data.",
    "Deep Learning uses multiple layers of neural networks.",
    "Natural Language Processing helps computers understand human language."
]

doc_embeddings = model.encode(documents)

question = "What is machine learning?"

question_embedding = model.encode([question])

scores = cosine_similarity(question_embedding, doc_embeddings)[0]

best_index = scores.argmax()
context = documents[best_index]

print("Question:", question)
print("\nRetrieved Context:", context)
print("\nAnswer:", context)
```

```python
# Experiment 6: End-to-End RAG Pipeline

!pip install -q sentence-transformers

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

model = SentenceTransformer('all-MiniLM-L6-v2')

documents = [
    "Solar energy is a renewable energy source obtained from sunlight.",
    "Wind energy is produced using wind turbines.",
    "Hydropower generates electricity using flowing water.",
    "Coal is a non-renewable source of energy.",
    "Geothermal energy uses heat from inside the Earth."
]

# Document chunking
chunks = documents

# Generate embeddings
embeddings = model.encode(chunks)

# User question
question = "Which energy source uses sunlight?"

# Retrieve relevant document
query_embedding = model.encode([question])
scores = cosine_similarity(query_embedding, embeddings)[0]

top_index = scores.argmax()
retrieved_context = chunks[top_index]

# Generate a simple extractive answer
answer = retrieved_context

print("Question:", question)
print("\nRetrieved Document:", retrieved_context)
print("\nGenerated Answer:", answer)
```

```python
# Experiment 7: Domain-Specific Chatbot Using LangChain and FAISS

!pip install -q langchain langchain-community langchain-text-splitters langchain-huggingface faiss-cpu sentence-transformers

from langchain_community.vectorstores import FAISS
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings

documents = [
    Document(page_content="Mechanical engineering deals with machines and mechanical systems."),
    Document(page_content="Electrical engineering deals with electrical circuits and power systems."),
    Document(page_content="Civil engineering focuses on buildings, roads and bridges."),
    Document(page_content="Computer engineering combines hardware and software."),
    Document(page_content="Chemical engineering deals with chemical processes.")
]

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

db = FAISS.from_documents(documents, embeddings)

while True:
    question = input("\nAsk an engineering question (type exit to stop): ")

    if question.lower() == "exit":
        break

    results = db.similarity_search(question, k=2)

    print("\nChatbot Response:")
    for doc in results:
        print("-", doc.page_content)
```

```python
# Experiment 8: Context-Aware Chatbot Using Retrieval

!pip install -q sentence-transformers

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

model = SentenceTransformer('all-MiniLM-L6-v2')

documents = [
    "Python is a high-level programming language.",
    "Python supports object-oriented programming.",
    "Python is widely used in artificial intelligence.",
    "Java is used for enterprise application development.",
    "C++ is commonly used in game development."
]

embeddings = model.encode(documents)

chat_history = []

while True:
    question = input("\nYou: ")

    if question.lower() == "exit":
        break

    query_embedding = model.encode([question])
    scores = cosine_similarity(query_embedding, embeddings)[0]

    index = scores.argmax()
    answer = documents[index]

    chat_history.append(("You", question))
    chat_history.append(("Bot", answer))

    print("Bot:", answer)
```

```python
# Experiment 9: AI Assistant Using External Documents

!pip install -q sentence-transformers pypdf

from google.colab import files
from pypdf import PdfReader
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

uploaded = files.upload()

text = ""

for filename in uploaded:
    reader = PdfReader(filename)

    for page in reader.pages:
        text += page.extract_text() or ""

documents = [line.strip() for line in text.split("\n") if len(line.strip()) > 30]

if not documents:
    print("No readable text found in the PDF.")
else:
    model = SentenceTransformer('all-MiniLM-L6-v2')

    embeddings = model.encode(documents)

    question = input("Ask a question about the uploaded document: ")

    query_embedding = model.encode([question])
    scores = cosine_similarity(query_embedding, embeddings)[0]

    top_indices = scores.argsort()[-3:][::-1]

    print("\nAI Assistant Response:")
    for idx in top_indices:
        print("-", documents[idx])
```

```python
# Experiment 10: Multi-Document AI Assistant

!pip install -q sentence-transformers pypdf

from google.colab import files
from pypdf import PdfReader
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

uploaded = files.upload()

documents = []

for filename in uploaded:
    reader = PdfReader(filename)

    for page in reader.pages:
        page_text = page.extract_text() or ""

        for line in page_text.split("\n"):
            if len(line.strip()) > 30:
                documents.append(line.strip())

if not documents:
    print("No readable text found in the uploaded documents.")
else:
    model = SentenceTransformer('all-MiniLM-L6-v2')

    embeddings = model.encode(documents)

    while True:
        question = input("\nAsk a question (type exit to stop): ")

        if question.lower() == "exit":
            break

        query_embedding = model.encode([question])

        scores = cosine_similarity(query_embedding, embeddings)[0]

        top_indices = scores.argsort()[-3:][::-1]

        print("\nRelevant Information:")

        for idx in top_indices[::-1]:
            print("-", documents[idx])
```
